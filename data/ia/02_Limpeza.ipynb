```python
df_limpo = df.copy()

print("Cópia criada!")
print(df_limpo.shape)
```

```python
(df_limpo['num_caracteres'] < 50).sum()
```

```python
df_limpo = df_limpo[df_limpo['num_caracteres'] >= 50]

print("Limpeza aplicada!")
print(df_limpo.shape)
```

```python
df_limpo.isnull().sum()
```

```python
df_limpo['conteudo_repetido'] = df_limpo['content'].duplicated(keep=False)

df_limpo['conteudo_repetido'].value_counts()
```

```python
df_limpo['collected_at'] = pd.to_datetime(df_limpo['collected_at'])

df_limpo.info()
```

```python
def classificar_documento(nome):
    if nome == "README.md":
        return "Documentacao"
    elif nome == "LICENSE":
        return "Licenca"
    elif nome in ["requirements.txt", "package.json", "Gemfile"]:
        return "Dependencias"
    elif nome in ["pom.xml", "build.gradle"]:
        return "Configuracao_Java"
    elif nome in ["go.mod", "Cargo.toml"]:
        return "Configuracao_Linguagem"
    else:
        return "Outro"

df_limpo['categoria_documento'] = df_limpo['file_name'].apply(classificar_documento)

df_limpo['categoria_documento'].value_counts()
```

```python
def classificar_documento(nome):
    if nome == "README.md":
        return "Documentacao"
    elif nome == "LICENSE":
        return "Licenca"
    elif nome in ["requirements.txt", "package.json", "Gemfile"]:
        return "Dependencias"
    elif nome in ["pom.xml", "build.gradle"]:
        return "Configuracao_Java"
    elif nome in ["go.mod", "Cargo.toml"]:
        return "Configuracao_Linguagem"
    elif nome == "CONTRIBUTING.md":
        return "Contribuicao"
    elif nome == "CHANGELOG.md":
        return "Historico_Alteracoes"
    elif nome == "CODE_OF_CONDUCT.md":
        return "Governanca"
    elif nome == "SECURITY.md":
        return "Seguranca"
    elif nome == "pyproject.toml":
        return "Configuracao_Python"
    else:
        return "Outro"


df_limpo['categoria_documento'] = df_limpo['file_name'].apply(classificar_documento)

df_limpo['categoria_documento'].value_counts()
```

```python
df_limpo[['file_name', 'categoria_documento']].sample(15)
```

```python
caminho_saida = "/content/drive/MyDrive/Hackathon_ONE/dataset/tratado/documentation_tratado.parquet"

df_limpo.to_parquet(caminho_saida, index=False)

print("Dataset tratado salvo com sucesso!")
```
