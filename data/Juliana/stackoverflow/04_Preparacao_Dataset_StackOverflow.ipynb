# 📌 Etapa 8 - Construção do Dataset do Projeto

## Objetivo

Nesta etapa será iniciado o processo de transformação do dataset original do StackExchange StackOverflow para o formato adotado pelo projeto SCRIPTO.

Como o dataset original possui apenas três campos principais (`instruction`, `completion` e `conversations`), serão criadas novas colunas para atender ao padrão definido pela equipe de desenvolvimento.

Nesta primeira transformação serão gerados os campos básicos de identificação e organização dos registros, mantendo todas as informações originais do dataset.

Essa etapa constitui a base para as próximas fases de enriquecimento e classificação automática dos conteúdos.

```python
# ==========================================================
# Construindo o novo DataFrame no padrão do projeto
# ==========================================================

import pandas as pd
import numpy as np

novo_df = pd.DataFrame()

# Identificador único do registro
novo_df["record_id"] = range(1, len(df) + 1)

# Identificador da fonte
novo_df["source_id"] = [
    f"SRC-{str(i).zfill(5)}"
    for i in range(1, len(df) + 1)
]

# Origem do conteúdo
novo_df["source_type"] = "stackoverflow"

# Mantendo os campos originais
novo_df["title"] = df["instruction"]
novo_df["content"] = df["completion"]
```

```python
# Visualizando a estrutura inicial

novo_df.head()
```

# 📌 Etapa 9 - Inclusão dos Metadados

## Objetivo

Nesta etapa serão adicionados metadados básicos ao conjunto de dados.



```python
# ==========================================================
# Adicionando metadados básicos
# ==========================================================

import numpy as np

# Autor
novo_df["author"] = "Desconhecido"

# Identificador do autor
novo_df["author_id"] = [
    f"AUT-{str(i).zfill(4)}"
    for i in range(1, len(novo_df)+1)
]

# Categoria inicial
novo_df["category"] = "Tecnologia"

# Data fictícia (será melhorada posteriormente)
novo_df["created_at"] = "2025-01-01"
```

```python
novo_df.head()
```

# 📌 Etapa 10 - Enriquecimento dos Metadados

## Objetivo

Nesta etapa serão gerados metadados sintéticos para tornar o conjunto de dados mais próximo de uma base real.

Serão criadas datas distribuídas ao longo do tempo, autores fictícios e categorias relacionadas ao conteúdo técnico.

Essas informações enriquecem o dataset e permitem simular um ambiente mais próximo daquele encontrado em aplicações reais de gerenciamento de conhecimento.

```python
# ==========================================================
# Enriquecendo os metadados
# ==========================================================

import random

# Datas aleatórias entre 2022 e 2025
datas = pd.date_range("2022-01-01", "2025-01-01")

novo_df["created_at"] = np.random.choice(datas, size=len(novo_df))

# Lista de autores fictícios
autores = [
    "Marina Almeida",
    "Carlos Mendes",
    "Daniel Rocha",
    "Ana Souza",
    "Bruno Lima",
    "Fernanda Costa",
    "Lucas Oliveira",
    "Juliana Martins",
    "Gabriel Silva",
    "Patrícia Ramos"
]

novo_df["author"] = np.random.choice(autores, size=len(novo_df))
```

```python
novo_df.head()
```

## Objetivo

Após a criação dos metadados sintéticos, é importante verificar se o conjunto de dados está seguindo a estrutura esperada pelo projeto.

# 📌 Etapa 12 - Organização Final das Colunas

## Objetivo

Nesta etapa é realizada a organização da ordem das colunas do dataset, seguindo o padrão definido pela equipe do projeto.

Essa padronização facilita a integração do conjunto de dados com as próximas etapas de treinamento do modelo de Inteligência Artificial.

```python
# ==========================================================
# Organizando as colunas conforme padrão do projeto
# ==========================================================

novo_df = novo_df[
    [
        "record_id",
        "source_id",
        "source_type",
        "author_id",
        "author",
        "created_at",
        "title",
        "content",
        "category"
    ]
]

novo_df.head()
```
